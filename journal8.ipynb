```python
import matplotlib.pyplot as plt
import numpy as np
n = 100
x = np.linspace(0,1,n)
```

```python
def sin_plot(x):
    return np.sin(x)
  
def cos_plot(x):
    return np.cos(x)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2)
ax1.plot(x,sin_plot(x))
ax2.plot(x,cos_plot(x))

ax1.set_xlabel('x')
ax1.set_ylabel('sin(x)')
ax2.set_xlabel('x')
ax2.set_ylabel('cos(x)')

plt.tight_layout()

flag_save = True
fname = 'my_plots.pdf'
if flag_save:
    plt.savefig(fname)
```
