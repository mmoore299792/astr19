```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
shape, scale = 2, 2
random_numbers = np.random.gamma(2,67,1000)
np.random.seed(67)

plt.hist(random_numbers,bins=100,alpha=0.9,color='royalblue',edgecolor='midnightblue')
plt.xlabel('Gamma Random Variable')
plt.ylabel('N per bin')
plt.savefig('gamma.pdf')
plt.show()
```
