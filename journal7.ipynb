```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
n = 100
x = np.linspace(0,1,n)
def fun(x):
    return np.exp(x)
```

```python
y = fun(x)
def awesome_plot(x,y):
    flag_save = True 
    xlabel = 'Time [milliseconds]'
    ylabel = 'Awesomeness' 
    lcolor = 'springgreen' 
    fs = 14
    fname = 'awesome_plot.pdf'
    
    plt.figure()
    plt.xlabel(xlabel, fontsize = fs)
    plt.ylabel(ylabel, fontsize = fs)
    plt.plot(x,y, color = lcolor)
    if flag_save:
        plt.savefig(fname)
awesome_plot(x,y)
plt.show()
```
